```python
import torch
import matplotlib.pyplot as plt
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
```

# 3.4.1 Defining the Model

```python
class LRS(nn.Module):
    def __init__(self,num_inputs,lr,sigma=0.01):
        super().__init__()

        self.lr=lr

        self.w=nn.Parameter(torch.normal(0,sigma,(num_inputs,1)))
        self.b=nn.Parameter(torch.zeros(1))


    def forward(self,X):
        return X@self.w+self.b

    def loss(self,y_hat,y):
        l=(y_hat-y)**2/2
        return l.mean()
```

# 3.4.3 Optimization Algorithm

```python
class LRS(nn.Module):
    def __init__(self,num_inputs,sigma=0.01):
        super().__init__()

        self.lr=lr

        self.w=nn.Parameter(torch.normal(0,sigma,(num_inputs,1)))
        self.b=nn.Parameter(torch.zeros(1))


    def forward(self,X):
        return X@self.w+self.b

    def loss(self,y_hat,y):
        l=(y_hat-y)**2/2
        return l.mean()
#First example of SGD!!! :))

class SGD():
    def __init__(self,params,lr):
        self.params=list(params)
        self.lr=lr


    def step(self):
        with torch.no_grad():
            for param in self.params:
                if param.grad is not None:
                    param-=self.lr*param.grad



    def zero_grad(self):
        for param in self.params:
            if param.grad is not None:
                param.grad.zero_() 

#DATA PREPARATION

true_w=torch.tensor([2.0,-3.4])
true_b=4.2

n=1000
X=torch.randn(1000,len(true_w))
y=X@true_w.reshape(-1,1)+true_b
y+=torch.randn(y.shape)*0.01

dataset=TensorDataset(X,y)
data_loader=DataLoader(dataset,batch_size=32,shuffle=True)

lr=0.03
model=LRS(num_inputs=2)
optimizer=SGD(model.parameters(),lr=lr)

n_epochs=9

for e in range(n_epochs):
    for batch_X,batch_y in data_loader:
        optimizer.zero_grad()
        y_hat=model(batch_X)
        l=model.loss(y_hat,batch_y)
        l.backward()
        optimizer.step()

    print(f"Epoch {e+1} | Loss: {l.item(): .6f}")



print("True weights:",true_w)
print("Predicted weights: ",model.w.data.squeeze())
print("-"*10)
print("True bias",true_b)
print("Predicted",model.b.data.item())

```
