# 3.1.2 Vectorization for Speed

```python
# When training our models, we usually want to process minibatches simultaneously.
# Doing this efficiently requires vectorization to avoid for-loops.

# Why this matters:
n=10000
a=torch.ones(n)
b=torch.ones(n)

# benchmark workloads:

c=torch.zeros(n)
t=time.time()
for i in range (n):
    c[i]=a[i]+b[i]

print(f"{time.time()-t}")

#alternatively:
t=time.time()
d=a+b
print(f"{time.time()-t}")

# second method is dramatically faster than the 1st. moreover, we push more of the maths to the library!!
```
