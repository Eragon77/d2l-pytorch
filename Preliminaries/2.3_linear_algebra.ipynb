# 2.3.1 Scalars

```python
import torch

#scalars are implemented by single-value tensors

x=torch.tensor(3.0)
y=torch.tensor(5.0)

print(x+y)
```

# 2.3.2 Vectors

```python
#denoted by bold lowercase letters (e.g. bold(x))

x=torch.arange(3)
print(x)
# we access the tensor via indexing.
print(x[1])

print(x.shape)
```

# 2.3.3 Matrices    

```python
#2nd order tensors
A=torch.arange(6).reshape(3,2)
print(A)

#transp:
print(A.T)

#symmetric: A=A^T, with A square matrix (m x n where m=n)
A=torch.zeros(4,4)

if torch.equal(A,A.T):
    print("A is symmetrical")
else:
    print("A is NOT symmetrical")

```

# 2.3.5 Basic Properties of Tensors Arithmetic

```python
#elementwise ops produce outputs that have the same size as their operands.

A=torch.arange(6,dtype=torch.float32).reshape(2,3)
B=A.clone() #allocation of new memory
print(A)
print(A+B)

#elementwise product is called Hadamard product O.
print(A*B)

#scalar+Tensor v scalar*Tensor do not change Tensor's shape.
a=2
X=torch.arange(24).reshape(2,3,4)
print(2*X)
```

# 2.3.6 Reduction

```python
#sum of tensor's elements:
x=torch.arange(3)
print(x.sum())
```
