# 2.1.5 Saving Memory

```python
before = id(Y)
Y = Y + X
r=id(Y) == before
print(r)
# so we deference the previous Y but we allocate use a fresh new space, allocating further (unnecessary) memory.
# therefore, we want to perform this updates IN PLACE, like so:
before=id(Y)
Y[:]=Y+X #or Y+=X
r=id(Y)==before 
print(r)
```
